```python
if not cached:
    import numpy as np

    def my_cov(matrix):
        cov_matrix = np.cov(matrix, rowvar=False)
        cov_svd = np.linalg.svd(cov_matrix, compute_uv=False)
        return np.sum(cov_svd)

    # def my_cov(matrix):
    #     cov_matrix = np.cov(matrix, rowvar=False)
    #     return np.linalg.norm(cov_matrix, ord='fro')

    def process(hidden, encoder, decoder):
        hidden = np.array(hidden)
        hidden = np.matmul(hidden, encoder.T)
        hidden = np.matmul(hidden, decoder.T)
        return hidden
```

```python
from sklearn.decomposition import PCA
import numpy as np
import scipy.linalg
from tqdm import tqdm

if not cached:
    cos_res_ins_clean = []
    cos_res_ins_w_label_space = []
    zero_shot_cos_res = []
    eight_shot_res = []
    eight_shot_random_label_res = []
    eight_shot_ood_res = []

    cos_res_ins_clean_ecce = []
    cos_res_ins_w_label_space_ecce = []
    zero_shot_cos_res_ecce = []
    eight_shot_res_ecce = []
    eight_shot_random_label_res_ecce = []
    eight_shot_ood_res_ecce = []

    cov_res_ins_clean = []
    cov_res_ins_w_label_space = []
    zero_shot_cov_res = []
    eight_shot_cov_res = []
    cov_res_eight_shot_random_label = []
    cov_res_eight_shot_ood = []

    layers = []

    for layer_index in tqdm(range(0, model.config.num_hidden_layers + 1, layer_jump)):
        # 如果 layer_index 不存在于 pkl_file_in_layer 的 keys 中，则跳过
        if layer_index not in pkl_file_in_layer:
            continue
        encoder = pkl_file_in_layer[layer_index]['auto_encoder_state_dict']['encoder.weight'].numpy()
        decoder = pkl_file_in_layer[layer_index]['auto_encoder_state_dict']['decoder.weight'].numpy()

        clean_ins_samples = []
        zero_shot = []
        label_space_ins_samples = []
        eight_shot_samples = []
        eight_shot_random_label_samples = []
        eight_shot_ood_samples = []

        for sample_index in range(len(hidden_state_data['post_test_res']['hidden_states'])):
            zero_shot.append(zero_shot_hidden_state_data['post_test_res']['hidden_states'][sample_index][layer_index][0][-1])
            clean_ins_samples.append(instruction_hidden_state_data_clean['post_test_res']['hidden_states'][sample_index][layer_index][0][-1])
            label_space_ins_samples.append(instruction_hidden_state_data_w_label_space['post_test_res']['hidden_states'][sample_index][layer_index][0][-1])
            eight_shot_samples.append(eight_shot_data['post_test_res']['hidden_states'][sample_index][layer_index][0][-1])
            eight_shot_random_label_samples.append(eight_shot_w_random_label_data['post_test_res']['hidden_states'][sample_index][layer_index][0][-1])
            eight_shot_ood_samples.append(eight_shot_ood_data['post_test_res']['hidden_states'][sample_index][layer_index][0][-1])

        clean_ins_samples_np = np.array(clean_ins_samples)
        zero_shot_np = np.array(zero_shot)
        label_space_ins_samples_np = np.array(label_space_ins_samples)
        eight_shot_samples_np = np.array(eight_shot_samples)
        eight_shot_random_label_samples_np = np.array(eight_shot_random_label_samples)
        eight_shot_ood_samples_np = np.array(eight_shot_ood_samples)

        zero_shot_np_processed = process(zero_shot_np, encoder, decoder)
        clean_ins_samples_np_processed = process(clean_ins_samples_np, encoder, decoder)
        label_space_ins_samples_np_processed = process(label_space_ins_samples_np, encoder, decoder)
        eight_shot_samples_np_processed = process(eight_shot_samples_np, encoder, decoder)
        eight_shot_random_label_samples_np_processed = process(eight_shot_random_label_samples_np, encoder, decoder)
        eight_shot_ood_samples_np_processed = process(eight_shot_ood_samples_np, encoder, decoder)

        zero_shot_np_processed_cov = my_cov(zero_shot_np_processed)
        clean_ins_samples_np_processed_cov = my_cov(clean_ins_samples_np_processed)
        label_space_ins_samples_np_processed_cov = my_cov(label_space_ins_samples_np_processed)
        eight_shot_samples_np_processed_cov = my_cov(eight_shot_samples_np_processed)
        eight_shot_random_label_samples_np_processed_cov = my_cov(eight_shot_random_label_samples_np_processed)
        eight_shot_ood_samples_np_processed_cov = my_cov(eight_shot_ood_samples_np_processed)

        zero_shot_original_cov = my_cov(zero_shot_np)
        clean_ins_samples_original_cov = my_cov(clean_ins_samples_np)
        label_space_ins_samples_original_cov = my_cov(label_space_ins_samples_np)
        eight_shot_samples_original_cov = my_cov(eight_shot_samples_np)
        eight_shot_random_label_samples_original_cov = my_cov(eight_shot_random_label_samples_np)
        eight_shot_ood_samples_original_cov = my_cov(eight_shot_ood_samples_np)

        cov_res_ins_clean.append(clean_ins_samples_np_processed_cov / clean_ins_samples_original_cov)
        cov_res_ins_w_label_space.append(label_space_ins_samples_np_processed_cov / label_space_ins_samples_original_cov)
        zero_shot_cov_res.append(zero_shot_np_processed_cov / zero_shot_original_cov)
        eight_shot_cov_res.append(eight_shot_samples_np_processed_cov / eight_shot_samples_original_cov)
        cov_res_eight_shot_random_label.append(eight_shot_random_label_samples_np_processed_cov / eight_shot_random_label_samples_original_cov)
        cov_res_eight_shot_ood.append(eight_shot_ood_samples_np_processed_cov / eight_shot_ood_samples_original_cov)

        PCA_clean = PCA(n_components=2)
        PCA_zero_shot = PCA(n_components=2)
        PCA_label_space = PCA(n_components=2)
        PCA_eight_shot = PCA(n_components=2)
        PCA_eight_shot_random_label = PCA(n_components=2)
        PCA_eight_shot_ood = PCA(n_components=2)

        scree_clean = np.cumsum(PCA_clean.fit(clean_ins_samples_np).explained_variance_ratio_)
        scree_zero_shot = np.cumsum(PCA_zero_shot.fit(zero_shot_np).explained_variance_ratio_)
        scree_label_space = np.cumsum(PCA_label_space.fit(label_space_ins_samples_np).explained_variance_ratio_)
        scree_eight_shot = np.cumsum(PCA_eight_shot.fit(eight_shot_samples_np).explained_variance_ratio_)
        scree_eight_shot_random_label = np.cumsum(PCA_eight_shot_random_label.fit(eight_shot_random_label_samples_np).explained_variance_ratio_)
        scree_eight_shot_ood = np.cumsum(PCA_eight_shot_ood.fit(eight_shot_ood_samples_np).explained_variance_ratio_)

        main_direction_clean = PCA_clean.components_[0]
        main_direction_zero_shot = PCA_zero_shot.components_[0]
        main_direction_label_space = PCA_label_space.components_[0]
        main_direction_eight_shot = PCA_eight_shot.components_[0]
        main_direction_eight_shot_random_label = PCA_eight_shot_random_label.components_[0]
        main_direction_eight_shot_ood = PCA_eight_shot_ood.components_[0]

        encoder = pkl_file_in_layer[layer_index]['auto_encoder_state_dict']['encoder.weight'].numpy()

        _0_shot_res = np.cos(scipy.linalg.subspace_angles(np.array([main_direction_zero_shot]).T, encoder.T))
        _ins_clean_res = np.cos(scipy.linalg.subspace_angles(np.array([main_direction_clean]).T, encoder.T))
        _ins_label_space_res = np.cos(scipy.linalg.subspace_angles(np.array([main_direction_label_space]).T, encoder.T))
        _eight_shot_res = np.cos(scipy.linalg.subspace_angles(np.array([main_direction_eight_shot]).T, encoder.T))
        _eight_shot_random_label_res = np.cos(scipy.linalg.subspace_angles(np.array([main_direction_eight_shot_random_label]).T, encoder.T))
        _eight_shot_ood_res = np.cos(scipy.linalg.subspace_angles(np.array([main_direction_eight_shot_ood]).T, encoder.T))

        cos_res_ins_clean.append(_ins_clean_res.item())
        cos_res_ins_w_label_space.append(_ins_label_space_res.item())
        zero_shot_cos_res.append(_0_shot_res.item())
        eight_shot_res.append(_eight_shot_res.item())
        eight_shot_random_label_res.append(_eight_shot_random_label_res.item())
        eight_shot_ood_res.append(_eight_shot_ood_res.item())

        eight_shot_res_ecce.append(scree_eight_shot[0])
        cos_res_ins_clean_ecce.append(scree_clean[0])
        cos_res_ins_w_label_space_ecce.append(scree_label_space[0])
        zero_shot_cos_res_ecce.append(scree_zero_shot[0])
        eight_shot_random_label_res_ecce.append(scree_eight_shot_random_label[0])
        eight_shot_ood_res_ecce.append(scree_eight_shot_ood[0])

        layers.append(layer_index)
```
