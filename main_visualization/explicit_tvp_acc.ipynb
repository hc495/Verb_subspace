```python
path = "logs/Qwen-2.5-3B-Instruct/"
model_name = "Qwen-2.5-3B-Instruct"
source_prefix = "ICL_6_"
encoding_path_map = {
    "ICL_0_": "logs/ICL_dataset_Qwen_Qwen2.5-3B-Instruct_index_0_20251116_063151.pkl",
    "ICL_1_": "logs/ICL_dataset_Qwen_Qwen2.5-3B-Instruct_index_1_20251116_063730.pkl",
    "ICL_2_": "logs/ICL_dataset_Qwen_Qwen2.5-3B-Instruct_index_2_20251116_064312.pkl",
    "ICL_3_": "logs/ICL_dataset_Qwen_Qwen2.5-3B-Instruct_index_3_20251116_064853.pkl",
    "ICL_5_": "logs/ICL_dataset_Qwen_Qwen2.5-3B-Instruct_index_5_20251116_065444.pkl",
    "ICL_6_": "logs/ICL_dataset_Qwen_Qwen2.5-3B-Instruct_index_6_20251116_070021.pkl"
}
encoding_path = encoding_path_map[source_prefix]

# additional_path = "../Llama-3.2-1B/" if source_prefix == "ICL_0_" else None
additional_path = None
model_dim = 2048

hook_layer = 8
hook_rank = 4
```

```python
import os
import pickle

dataset = []
# 遍历path中含有source_prefix的所有文件夹
for folder in os.listdir(path):
    folder_path = os.path.join(path, folder)
    if os.path.isdir(folder_path) and source_prefix in folder:
        # 遍历文件夹中的所有pkl文件
        for file in os.listdir(folder_path):
            if file.endswith('.pkl'):
                file_path = os.path.join(folder_path, file)
                with open(file_path, 'rb') as f:
                    try:
                        data = pickle.load(f)
                    except Exception as e:
                        print(f"Corrupted file skipped: {file_path}, error: {e}")
                        continue
                dataset.append(data)
                # 这里可以对data进行处理
                print(f"Loaded {file_path}")
```

```python
if additional_path:
    for folder in os.listdir(additional_path):
        folder_path = os.path.join(additional_path, folder)
        if os.path.isdir(folder_path) and source_prefix in folder:
            for file in os.listdir(folder_path):
                if file.endswith('.pkl'):
                    file_path = os.path.join(folder_path, file)
                    with open(file_path, 'rb') as f:
                        data = pickle.load(f)
                    dataset.append(data)
                    print(f"Loaded {file_path} from additional_path")
```

```python
hook_res = []
injected_results = {}
for data in dataset:
    if data['args'].injected_rank not in injected_results:
        injected_results[data['args'].injected_rank] = {}
    if data['args'].injected_layer_num not in injected_results[data['args'].injected_rank]:
        injected_results[data['args'].injected_rank][data['args'].injected_layer_num] = []
    if data['post_test_res'] is None:
        injected_results[data['args'].injected_rank][data['args'].injected_layer_num].append(0)
    else:
        injected_results[data['args'].injected_rank][data['args'].injected_layer_num].append(data['post_test_res']['res'])
    if hook_layer == data['args'].injected_layer_num and hook_rank == data['args'].injected_rank:
        hook_res.append(data['auto_encoder_state_dict'])
```
