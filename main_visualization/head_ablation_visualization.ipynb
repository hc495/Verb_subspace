```python
import glob
import os

def get_pkl_file(path):
    if path is None:
        return None
    if os.path.isdir(path):
        pkl_files = glob.glob(os.path.join(path, "*.pkl"))
        if not pkl_files:
            raise FileNotFoundError(f"No .pkl files found in directory {path}")
        return pkl_files[0]
    return path
```

```python
import numpy as np

def my_cov(matrix):
    cov_matrix = np.cov(matrix, rowvar=False)
    cov_svd = np.linalg.svd(cov_matrix, compute_uv=False)
    return np.sum(cov_svd)

# def my_cov(matrix):
#     cov_matrix = np.cov(matrix, rowvar=False)
#     return np.linalg.norm(cov_matrix, ord='fro')

def process(hidden, encoder, decoder):
    hidden = np.array(hidden)
    hidden = np.matmul(hidden, encoder.T)
    hidden = np.matmul(hidden, decoder.T)
    return hidden
```

```python
import pickle
from tqdm import tqdm
import os
from sklearn.decomposition import PCA

ecce_res = {}
cov_res = {}
acc_res = {}

if os.path.exists(final_save_path + '/' + save_name):
# if False:
    with open(final_save_path + '/' + save_name, 'rb') as f:
        data = pickle.load(f)
    ecce_res = data['ecce_res']
    cov_res = data['cov_res']
    acc_res = data['acc_res']
else:
    print("No existing data found, starting fresh.")
    for layer in range(start_layer, max_layer, layer_jump): 
        print(f"Processing layer {layer}...")
        layer_data_save_name = f"layer_{layer}_head_ablation_data.pickle"
        if os.path.exists(final_save_path + '/' + layer_data_save_name):
            with open(final_save_path + '/' + layer_data_save_name, 'rb') as f:
                layer_data = pickle.load(f)
            ecce_res[layer] = layer_data['ecce_res_layer']
            cov_res[layer] = layer_data['cov_res_layer']
            acc_res[layer] = layer_data['acc_res_layer']
            print(f"Layer {layer} data loaded from cache.")
            continue

        # 第0层实际上是第1层，对应第1个encoder。你总不可能ablate第0层的头吧？
        contains = 'layer_' + str(layer) 

        pkl_files = []
        for folder in tqdm(folder_list):
            if contains not in folder:
                continue
            for file in os.listdir(folder):
                if file.endswith('.pkl'):
                    file_path = os.path.join(folder, file)
                    with open(file_path, 'rb') as f:
                        pkl_data = pickle.load(f)
                        pkl_files.append(pkl_data)

        folder_prefix = folder_name
        folder_postfix = "_8" # rank

        # 获取zero_shot_log_path中的所有文件夹
        folders = [f for f in os.listdir(zero_shot_log_path) if os.path.isdir(os.path.join(zero_shot_log_path, f))]
        pkl_file_list = []
        for folder in folders:
            if folder.startswith(folder_prefix) and folder.endswith(folder_postfix):
                folder_path = os.path.join(zero_shot_log_path, folder)
                files = [f for f in os.listdir(folder_path) if f.endswith('.pkl')]
                if files:
                    pkl_path = os.path.join(folder_path, files[0])
                    with open(pkl_path, 'rb') as f:
                        pkl_file_list.append(pickle.load(f))

        pkl_file_in_layer = {}
        for pkl_file in pkl_file_list:
            pkl_file_in_layer[pkl_file['args'].injected_layer_num] = pkl_file

        selected_results = []

        for data in pkl_files:
            if data['args'].ablated_heads[layer] != []:
                selected_results.append(data)

        encoder = pkl_file_in_layer[layer + 1]['auto_encoder_state_dict']['encoder.weight'].numpy()
        decoder = pkl_file_in_layer[layer + 1]['auto_encoder_state_dict']['decoder.weight'].numpy()

        hidden_states_original = []
        for sample_index in range(len(raw_8_shot_data['post_test_res']['hidden_states'])):
            hidden_states_original.append(raw_8_shot_data['post_test_res']['hidden_states'][sample_index][layer + 1][0][-1])
        hidden_states_original = np.array(hidden_states_original)
        hidden_states_original_processed = process(hidden_states_original, encoder, decoder)
        pca = PCA(n_components=2)
        scree_clean = np.cumsum(pca.fit(hidden_states_original).explained_variance_ratio_)
        ecce_ori = scree_clean[0]
        cov_ori = my_cov(hidden_states_original_processed) / my_cov(hidden_states_original)

        ecce_res_layer = {}
        cov_res_layer = {}
        acc_res_layer = {}

        for data in tqdm(selected_results):
            hidden = []
            for sample_index in range(len(data['post_test_res']['hidden_states'])):
                hidden.append(data['post_test_res']['hidden_states'][sample_index][layer + 1][0][-1])
            hidden = np.array(hidden)
            hidden_processed = process(hidden, encoder, decoder)

            hidden_cov = my_cov(hidden)
            hidden_processed_cov = my_cov(hidden_processed)

            pca_hidden = PCA(n_components=2)
            scree_clean = np.cumsum(pca_hidden.fit(hidden).explained_variance_ratio_)

            # if data['args'].ablated_heads[layer][0] == 15 and layer == 9:
            #     print(hidden_processed_cov / hidden_cov)
            #     print(cov_ori)

            ecce_res_layer[data['args'].ablated_heads[layer][0]] = (scree_clean[0] - ecce_ori) / ecce_ori
            cov_res_layer[data['args'].ablated_heads[layer][0]] = (hidden_processed_cov / hidden_cov - cov_ori) / cov_ori
            acc_res_layer[data['args'].ablated_heads[layer][0]] = (data['post_test_res']['res'] - raw_8_shot_data['post_test_res']['res']) / raw_8_shot_data['post_test_res']['res']

        ecce_res[layer] = ecce_res_layer
        cov_res[layer] = cov_res_layer
        acc_res[layer] = acc_res_layer

        layer_data = {
            'ecce_res_layer': ecce_res_layer,
            'cov_res_layer': cov_res_layer,
            'acc_res_layer': acc_res_layer
        }
        with open(final_save_path + '/' + layer_data_save_name, 'wb') as f:
            pickle.dump(layer_data, f)
            print(f"Layer {layer} data saved to cache.")
```

```python
from scipy.stats import spearmanr

plain_point_ecce = []
plain_point_acc = []
plain_point_cov = []
plain_point_layer = []

for layer in range(start_layer, max_layer, layer_jump):
    ecce_res_layer = ecce_res[layer]
    cov_res_layer = cov_res[layer]
    acc_res_layer = acc_res[layer]    
    
    cov_res_temp = []
    acc_res_temp = []
    for i in range(len(ecce_res_layer)):
        if ecce_res_layer[i] != 0 and cov_res_layer[i] != 0 and acc_res_layer[i] != 0:
            cov_res_temp.append(cov_res_layer[i])
            acc_res_temp.append(acc_res_layer[i])
    correlation, p_value = spearmanr(cov_res_temp, acc_res_temp)
    print(f"Layer {layer} - Spearman correlation between Covariance and Accuracy: {correlation:.4f}, p-value: {p_value:.4f}")

    for i in range(len(ecce_res_layer)):
        plain_point_ecce.append(ecce_res_layer[i])
        plain_point_acc.append(acc_res_layer[i])
        plain_point_cov.append(cov_res_layer[i])
        plain_point_layer.append(layer)
```

```python
induction_threshold = 0.35
covarience_thereshold = -0.05

induction_heads = {}
denoising_heads = {}
for layer in range(start_layer, max_layer, layer_jump):
    induction_res_layer = induction_res['mean_magnitude'][layer]
    cov_res_layer = cov_res[layer]
    for i in range(len(induction_res_layer)):
        if induction_res_layer[i] > induction_threshold:
            if layer not in induction_heads:
                induction_heads[layer] = []
            induction_heads[layer].append(i)
        elif cov_res_layer[i] < covarience_thereshold:
            if layer not in denoising_heads:
                denoising_heads[layer] = []
            denoising_heads[layer].append(i)
```

```python
# 计算denoising_heads和induction_heads的重合率
overlap_count = 0
total_denoising = 0

for layer in denoising_heads:
    denoise_set = set(denoising_heads[layer])
    induct_set = set(induction_heads.get(layer, []))
    overlap_count += len(denoise_set & induct_set)
    total_denoising += len(denoise_set)

if total_denoising > 0:
    overlap_ratio = overlap_count / total_denoising
else:
    overlap_ratio = 0

print(f"重合率: {overlap_ratio:.4f}")
```

```python
# 合并denoising_heads和induction_heads为一个新的字典 all_heads
all_heads = {}

for layer in range(max_layer):
    denoise = set(denoising_heads.get(layer, []))
    induct = set(induction_heads.get(layer, []))
    all_heads[layer] = sorted(list(denoise | induct))

print(all_heads)
```
